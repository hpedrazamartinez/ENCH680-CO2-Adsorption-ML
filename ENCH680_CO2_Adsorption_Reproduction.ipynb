```python
import warnings, json, os
warnings.filterwarnings("ignore")
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import (train_test_split, GridSearchCV, cross_val_score,
                                     KFold, GroupKFold, cross_validate)
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.inspection import permutation_importance, PartialDependenceDisplay
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
import sklearn, xgboost, lightgbm

SEED = 42
os.makedirs("figs", exist_ok=True)
plt.rcParams.update({"figure.dpi": 110, "font.size": 11, "axes.spines.top": False, "axes.spines.right": False})
print("scikit-learn", sklearn.__version__, "| xgboost", xgboost.__version__, "| lightgbm", lightgbm.__version__)
RESULTS = {}
```

## 2. Preprocessing (same as the paper's Supporting Information, Tables S3–S4)
TPV and MPV are strongly correlated with surface area (SA), so the authors imputed the missing values
with a random forest that uses SA as the only input (80/20 split). After that, S (%) is dropped, which leaves **9 input features**.

```python
SA = "Surface Area (m2/g)"; TPV = "Total Pore Volume(cm3/g)"; MPV = "Micropore Volume (cm3/g)"
TARGET = "CO2 uptake (mmol/g)"
df = raw.copy()
imp_rows = []
for col in [TPV, MPV]:
    known = df[[SA, col]].dropna()
    Xk, yk = known[[SA]].values, known[col].values
    Xtr, Xte, ytr, yte = train_test_split(Xk, yk, test_size=0.2, random_state=SEED)
    for name, m in [("Linear regression", LinearRegression()), ("Random forest", RandomForestRegressor(random_state=SEED))]:
        m.fit(Xtr, ytr)
        imp_rows.append([col, name, r2_score(ytr, m.predict(Xtr)), r2_score(yte, m.predict(Xte))])
    rf_imp = RandomForestRegressor(random_state=SEED).fit(Xtr, ytr)
    miss = df[col].isna()
    df.loc[miss, col] = rf_imp.predict(df.loc[miss, [SA]].values)
    print(f"{col}: imputed {miss.sum()} values")
imp = pd.DataFrame(imp_rows, columns=["feature", "model", "train R2", "test R2"]).round(3)
RESULTS["imputation"] = imp.to_dict(orient="records")
df = df.drop(columns=["S (%)"])
FEATURES = [c for c in df.columns if c != TARGET]
print("Features used:", FEATURES)
imp
```

## 3. Train/test split (85/15, `random_state=42`, same as the authors' code)

```python
df[FEATURES] = df[FEATURES].astype(float)
X, y = df[FEATURES], df[TARGET]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=SEED)
print("Train:", X_train.shape, "Test:", X_test.shape)
cv5 = KFold(n_splits=5, shuffle=True, random_state=SEED)

def evaluate(name, model, Xtr=None, ytr=None, Xte=None, yte=None, store=True):
    Xtr = X_train if Xtr is None else Xtr; ytr = y_train if ytr is None else ytr
    Xte = X_test if Xte is None else Xte; yte = y_test if yte is None else yte
    model.fit(Xtr, ytr)
    p_tr, p_te = model.predict(Xtr), model.predict(Xte)
    cv = cross_val_score(model, Xtr, ytr, cv=cv5, scoring="r2")
    row = {"model": name,
           "train R2": r2_score(ytr, p_tr),
           "5-fold CV R2": cv.mean(), "CV R2 std": cv.std(),
           "test R2": r2_score(yte, p_te),
           "test RMSE": np.sqrt(mean_squared_error(yte, p_te)),
           "test MAE": mean_absolute_error(yte, p_te)}
    return row, model, p_tr, p_te
```

## 4. Chapter 3 baselines: decision tree and random forest
Chapter 3 shows that a single tree with no limits keeps splitting until each leaf is pure, which leads to overfitting.
Pre-pruning (`max_depth`, `min_samples_split`, `min_samples_leaf`) controls this. A random forest averages many
bootstrapped, decorrelated trees to reduce variance.

```python
rows, models, preds = [], {}, {}

r, m, ptr, pte = evaluate("Decision tree (unpruned)", DecisionTreeRegressor(random_state=SEED))
rows.append(r); models[r["model"]] = m; preds[r["model"]] = (ptr, pte)
print("Unpruned tree depth:", m.get_depth(), "| leaves:", m.get_n_leaves())

dt_grid = GridSearchCV(DecisionTreeRegressor(random_state=SEED),
                       {"max_depth": [3, 4, 5, 6, 8, 10, None],
                        "min_samples_split": [2, 5, 10, 20],
                        "min_samples_leaf": [1, 2, 4, 8]}, cv=cv5, scoring="r2", n_jobs=-1).fit(X_train, y_train)
print("Pre-pruned tree best params:", dt_grid.best_params_)
r, m, ptr, pte = evaluate("Decision tree (pre-pruned)", DecisionTreeRegressor(random_state=SEED, **dt_grid.best_params_))
rows.append(r); models[r["model"]] = m; preds[r["model"]] = (ptr, pte)

rf_grid = GridSearchCV(RandomForestRegressor(random_state=SEED, n_jobs=-1),
                       {"n_estimators": [200, 500], "max_depth": [None, 10, 6],
                        "max_features": [1.0, 0.6, 0.33], "min_samples_leaf": [1, 2]},
                       cv=cv5, scoring="r2", n_jobs=-1).fit(X_train, y_train)
print("Random forest best params:", rf_grid.best_params_)
r, m, ptr, pte = evaluate("Random forest (tuned)", RandomForestRegressor(random_state=SEED, n_jobs=-1, **rf_grid.best_params_))
rows.append(r); models[r["model"]] = m; preds[r["model"]] = (ptr, pte)
RESULTS["dt_params"] = dt_grid.best_params_; RESULTS["rf_params"] = rf_grid.best_params_
pd.DataFrame(rows).round(3)
```

## 5. Reproducing the paper: GBDT, XGBoost, LightGBM
**5a. With the authors' final hyperparameters** (taken from their published notebook).
**5b. With our own grid search** (5-fold CV on the training set), as the paper describes.

```python
paper_params = {
    "GBDT": GradientBoostingRegressor(learning_rate=0.05, max_depth=4, min_samples_leaf=1, min_samples_split=2,
                                      n_estimators=300, subsample=0.5, random_state=SEED),
    "XGB": XGBRegressor(colsample_bytree=1, learning_rate=0.05, max_depth=4, n_estimators=100,
                        subsample=0.5, random_state=SEED, eval_metric="rmse"),
    "LGB": LGBMRegressor(boosting_type="dart", colsample_bytree=1, learning_rate=0.2, max_depth=4,
                         n_estimators=300, num_leaves=5, reg_lambda=0.8, subsample=1, random_state=SEED, verbose=-1),
}
for k, m in paper_params.items():
    r, m, ptr, pte = evaluate(f"{k} (authors' hyperparameters)", m)
    rows.append(r); models[r["model"]] = m; preds[r["model"]] = (ptr, pte)
pd.DataFrame(rows).round(3).tail(3)
```

```python
grids = {
    "GBDT": (GradientBoostingRegressor(random_state=SEED),
             {"n_estimators": [100, 300, 500], "learning_rate": [0.02, 0.05, 0.1],
              "max_depth": [3, 4, 5], "subsample": [0.5, 0.8, 1.0]}),
    "XGB": (XGBRegressor(random_state=SEED, eval_metric="rmse", n_jobs=1),
            {"n_estimators": [100, 300, 500], "learning_rate": [0.02, 0.05, 0.1],
             "max_depth": [3, 4, 5], "subsample": [0.5, 0.8, 1.0]}),
    "LGB": (LGBMRegressor(random_state=SEED, verbose=-1, n_jobs=1),
            {"n_estimators": [100, 300, 500], "learning_rate": [0.02, 0.05, 0.1],
             "num_leaves": [5, 15, 31], "min_child_samples": [5, 10, 20]}),
}
tuned_params = {}
for k, (est, g) in grids.items():
    gs = GridSearchCV(est, g, cv=cv5, scoring="r2", n_jobs=-1).fit(X_train, y_train)
    tuned_params[k] = gs.best_params_
    print(k, "best params:", gs.best_params_)
    r, m, ptr, pte = evaluate(f"{k} (our grid search)", est.set_params(**gs.best_params_))
    rows.append(r); models[r["model"]] = m; preds[r["model"]] = (ptr, pte)
RESULTS["tuned_params"] = tuned_params
results = pd.DataFrame(rows).set_index("model")
RESULTS["main_table"] = results.round(4).reset_index().to_dict(orient="records")
results.round(3)
```

## 7. Model interpretation: permutation importance (MDA) and partial dependence
The paper uses permutation importance (mean decrease in accuracy). Chapter 3 warns that impurity-based importance
can be misleading when features are correlated (SA, TPV, MPV), so permutation importance on the **test set** is the safer choice.

```python
gbdt = models[best]
pi = permutation_importance(gbdt, X_test, y_test, n_repeats=30, random_state=SEED, scoring="r2")
imp_df = pd.DataFrame({"feature": FEATURES, "MDA (drop in R²)": pi.importances_mean, "std": pi.importances_std,
                       "impurity importance": gbdt.feature_importances_}).sort_values("MDA (drop in R²)", ascending=True)
RESULTS["importance"] = imp_df.round(4).to_dict(orient="records")
fig, ax = plt.subplots(figsize=(8, 4.8))
ax.barh(imp_df["feature"], imp_df["MDA (drop in R²)"], xerr=imp_df["std"], color="#1F4E79")
ax.set_xlabel("Mean decrease in test R² when feature is shuffled"); ax.set_title("GBDT permutation feature importance")
plt.tight_layout(); plt.savefig("figs/fig4_permutation_importance.png", dpi=200); plt.show()
imp_df.sort_values("MDA (drop in R²)", ascending=False).round(3)
```

```python
pdp_feats = [SA, MPV, TPV, "N (%)", "O (%)", "Temp (°C)", "Pressure (bar)", "C (%)"]
fig, ax = plt.subplots(2, 4, figsize=(16, 7))
PartialDependenceDisplay.from_estimator(gbdt, X_train, pdp_feats, ax=ax.ravel(), line_kw={"color": "#1F4E79"})
fig.suptitle("Partial dependence of CO₂ uptake (mmol/g) on each input feature (GBDT)")
plt.tight_layout(); plt.savefig("figs/fig5_pdp.png", dpi=200); plt.show()
```

## 9. Model improvements and robustness checks
### 9a. How much does the result depend on one random split?
The paper reports a single 85/15 split. Here we repeat the split with 50 different seeds.

```python
seed_scores = {"GBDT": [], "Random forest": []}
for s in range(50):
    a, b, c, d = train_test_split(X, y, test_size=0.15, random_state=s)
    seed_scores["GBDT"].append(r2_score(d, sklearn.base.clone(paper_params["GBDT"]).fit(a, c).predict(b)))
    seed_scores["Random forest"].append(r2_score(d, RandomForestRegressor(random_state=SEED, n_jobs=-1, **rf_grid.best_params_).fit(a, c).predict(b)))
ss = pd.DataFrame(seed_scores)
RESULTS["seed_summary"] = ss.describe().round(4).to_dict()
fig, ax = plt.subplots(figsize=(8, 4.2))
ax.hist(ss["GBDT"], bins=15, color="#1F4E79", alpha=0.8, label="GBDT")
ax.hist(ss["Random forest"], bins=15, color="#E8A33D", alpha=0.6, label="Random forest")
ax.axvline(0.84, color="red", ls="--", label="Paper GBDT test R² = 0.84")
ax.set_xlabel("Test R² (50 random 85/15 splits)"); ax.set_ylabel("Count"); ax.legend()
ax.set_title("Sensitivity of test R² to the random train/test split")
plt.tight_layout(); plt.savefig("figs/fig6_split_sensitivity.png", dpi=200); plt.show()
ss.describe().round(3)
```

### 9b. Data leakage check: grouped cross-validation by material
Many adsorbents in the dataset appear several times, measured at different temperatures (0 °C, 25 °C) and pressures (0.15 bar, 1 bar).
With a random split, the same material can be in both the training set and the test set, so the model is partly "remembering" the material.
To test a truly **new** material, we group rows by their material descriptors (SA, TPV, MPV, C, H, N, O) and use `GroupKFold`.

```python
mat_cols = [SA, TPV, MPV, "C (%)", "H (%)", "N (%)", "O (%)"]
groups = df[mat_cols].round(4).astype(str).agg("|".join, axis=1)
groups = pd.factorize(groups)[0]
print(f"{len(df)} rows -> {groups.max()+1} unique materials; "
      f"{(pd.Series(groups).value_counts() > 1).sum()} materials appear more than once")
gk = GroupKFold(n_splits=5)
kf = KFold(n_splits=5, shuffle=True, random_state=SEED)
leak_rows = []
for nm, est in [("GBDT", paper_params["GBDT"]), ("Random forest", RandomForestRegressor(random_state=SEED, n_jobs=-1, **rf_grid.best_params_)),
                ("Decision tree (pre-pruned)", DecisionTreeRegressor(random_state=SEED, **dt_grid.best_params_))]:
    r_kf = cross_validate(sklearn.base.clone(est), X, y, cv=kf, scoring=("r2", "neg_root_mean_squared_error"))
    r_gk = cross_validate(sklearn.base.clone(est), X, y, cv=gk, groups=groups, scoring=("r2", "neg_root_mean_squared_error"))
    leak_rows.append({"model": nm, "random 5-fold R2": r_kf["test_r2"].mean(), "grouped 5-fold R2": r_gk["test_r2"].mean(),
                      "random RMSE": -r_kf["test_neg_root_mean_squared_error"].mean(),
                      "grouped RMSE": -r_gk["test_neg_root_mean_squared_error"].mean()})
leak = pd.DataFrame(leak_rows)
RESULTS["leakage"] = leak.round(4).to_dict(orient="records")
RESULTS["n_materials"] = int(groups.max() + 1)
leak.round(3)
```

### 9c. SHAP explanation of the GBDT model
SHAP shows the direction of each feature's effect for every single prediction, which permutation importance cannot show.

```python
import shap
explainer = shap.TreeExplainer(gbdt)
sv = explainer.shap_values(X_test)
plt.figure()
shap.summary_plot(sv, X_test, show=False, plot_size=(9, 5))
plt.title("SHAP summary (GBDT, test set)")
plt.tight_layout(); plt.savefig("figs/fig7_shap.png", dpi=200); plt.show()
```

### 9d. "No extrapolation" (Chapter 3): trees predict a flat value outside the training range
Pressure in the dataset only goes from 0.10 to 1.01 bar. Physically, CO₂ uptake keeps rising with pressure
(Langmuir-type isotherm), but a tree model cannot predict this beyond the training data.

```python
ref = X_train.median().to_frame().T
pgrid = np.linspace(0.05, 3.0, 120)
Xp = pd.concat([ref] * len(pgrid), ignore_index=True); Xp["Pressure (bar)"] = pgrid
fig, ax = plt.subplots(figsize=(8, 4.2))
for t, col in [(0, "#1F4E79"), (25, "#E8A33D")]:
    Xp["Temp (°C)"] = t
    ax.plot(pgrid, gbdt.predict(Xp), color=col, lw=2, label=f"GBDT prediction, {t} °C")
ax.axvspan(X["Pressure (bar)"].min(), X["Pressure (bar)"].max(), color="grey", alpha=0.15, label="Pressure range in training data")
ax.set_xlabel("Pressure (bar)"); ax.set_ylabel("Predicted CO₂ uptake (mmol/g)")
ax.set_title("Tree models do not extrapolate: prediction is flat above 1 bar"); ax.legend()
plt.tight_layout(); plt.savefig("figs/fig8_extrapolation.png", dpi=200); plt.show()
```
